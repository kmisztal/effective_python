## 5. Example

#### The example of how simple and elegant it will be

````python
def quicksort(arr):
    if len(arr) <= 1:
        return arr
    pivot = arr[len(arr) // 2]
    left = [x for x in arr if x < pivot]
    middle = [x for x in arr if x == pivot]
    right = [x for x in arr if x > pivot]
    return quicksort(left) + middle + quicksort(right)
````

````python
print(quicksort([3,6,8,10,1,2,1, 4555]))
````

### Co znaczy „efektywny Python”?

- **Czytelność**: łatwo zrozumieć i utrzymać rozwiązanie.
- **Poprawność**: wynik jest zgodny z wymaganiami, także dla przypadków brzegowych.
- **Dobór algorytmu i narzędzi**: używaj sprawdzonych funkcji, gdy rozwiązują problem.
- **Wydajność**: mierz czas i zużycie pamięci, zanim zaczniesz optymalizować.

Ten quicksort pokazuje zwięzłość zapisu, ale tworzy dodatkowe listy i używa rekurencji. W zwykłym kodzie zacznij od `sorted()`. Krótszy kod nie musi być szybszy; najpierw sprawdź poprawność, potem wykonaj pomiar.

### Wspólna procedura: od poprawnego kodu do pomiaru

**Określ wymagania → sprawdź wynik i skutki uboczne → zmierz → zmień rozwiązanie → ponownie sprawdź poprawność → porównaj czas i pamięć.**

Przed porównaniem ustal dopuszczalne dane, wynik, zachowanie dla pustych lub brakujących danych i to, czy funkcja może zmieniać przekazane obiekty. Sama zgodność wyniku dla jednego przykładu nie wystarcza.

````python
numbers = [3, 6, 8, 10, 1, 2, 1, 4555]
print(sorted(numbers))
assert quicksort(numbers) == sorted(numbers)
````

#### Function with docs

````python
def gcd(a, b):
    """Zwróć nieujemny największy wspólny dzielnik dwóch liczb całkowitych.

    Dla gcd(0, 0) zwróć 0, podobnie jak math.gcd.
    """
    while b:
        a, b = b, a % b
    return abs(a)
````

````python
gcd_a, gcd_b = 1024, 5103
print(gcd(gcd_a, gcd_b))
````

#### ... or just find it and use it!

````python
from math import gcd as gcd_
print(gcd_(gcd_a, gcd_b))  # Te same argumenty co w naszej funkcji.
assert gcd(gcd_a, gcd_b) == gcd_(gcd_a, gcd_b)
````

### Jak działa `assert`?

Składnia: `assert warunek` lub `assert warunek, "komunikat"`.

- Python sprawdza wartość logiczną warunku. Jeśli jest prawdziwy, wykonanie przechodzi do następnej instrukcji — bez wypisywania wyniku.
- Jeśli warunek jest fałszywy, zgłasza wyjątek `AssertionError`. Opcjonalny komunikat wyjaśnia, co się nie zgadza; jest obliczany dopiero przy niepowodzeniu.
- Nieobsłużony wyjątek przerywa wykonywanie bieżącej komórki notebooka. Kolejne komórki nadal można uruchomić.

W ćwiczeniach asercje pomagają sprawdzić wynik i przypadki brzegowe przed pomiarem wydajności.

````python
assert 2 + 2 == 4, "Nieoczekiwany wynik dodawania"
print("Prawdziwy warunek: wykonanie trwa dalej")

# Przechwytujemy celowy błąd, aby pokaz wykonał się do końca.
try:
    assert 2 + 2 == 5, "Oczekiwano 5, ale wynik wynosi 4"
except AssertionError as error:
    print(f"AssertionError: {error}")
````

### Kiedy używać asercji?

Używaj `assert` do sprawdzania założeń programisty oraz prostych testów przykładów.

**Nie używaj go do walidacji danych użytkownika ani obsługi zwykłych błędów programu.** Uruchomienie Pythona z `-O` (również `-OO`) usuwa instrukcje `assert`: ani warunek, ani komunikat nie zostaną wtedy obliczone. Nie umieszczaj w nich operacji koniecznych do działania programu.

Dane wejściowe sprawdzaj jawnie, np. przez `if` i `raise ValueError`. Poniższe asercje sprawdzają poprawność naszej funkcji `gcd`.

````python
assert gcd(6, -4) == gcd_(6, -4) == 2
assert gcd(0, 0) == gcd_(0, 0) == 0
assert gcd(1024, 512) == 512
print("Przykładowe sprawdzenia gcd zakończone powodzeniem")
````

### `==` a `is` — wartość i tożsamość

- `==` porównuje wartości; dla list porównuje ich elementy.
- `is` sprawdza, czy obie nazwy wskazują **ten sam obiekt**.
- Do sprawdzania braku wartości używaj `value is None`.
- Do porównywania liczb i napisów używaj `==`; współdzielenie obiektów przez interpreter nie jest regułą, na której należy opierać takie porównania.

````python
first = [1, 2]
second = [1, 2]
same_object = first
print(first == second)       # True: te same wartości
print(first is second)       # False: dwa obiekty
print(first is same_object)  # True: ten sam obiekt

value = None
print(value is None)         # True
assert first == second and first is not second
assert same_object is first
````

### Dlaczego `0.1 + 0.2 != 0.3`?

`float` zapisuje liczby jako skończone przybliżenia binarne. Wartości takie jak `0.1` i `0.2` nie mają dokładnej skończonej reprezentacji w tym formacie; operacje mogą wprowadzać błędy zaokrąglenia.

Do porównywania przybliżonych wyników użyj `math.isclose()` i dobierz tolerancję do problemu. Przy porównaniu z zerem zwykle potrzebna jest dodatnia tolerancja bezwzględna `abs_tol`.

[Dokumentacja: arytmetyka zmiennoprzecinkowa](https://docs.python.org/3/tutorial/floatingpoint.html).

````python
import math

result = 0.1 + 0.2
print(result == 0.3)  # False
print(math.isclose(result, 0.3, rel_tol=1e-9, abs_tol=0.0))  # True
print(math.isclose(1e-12, 0.0, abs_tol=1e-9))  # True
assert math.isclose(result, 0.3, rel_tol=1e-9)
assert math.isclose(1e-12, 0.0, abs_tol=1e-9)
````

### Dopasowywanie wzorców — match/case (Python 3.10+)

`match` dopasowuje wartość do wzorców w kolejnych gałęziach `case`. Wykonuje się tylko pierwsza pasująca gałąź; nie ma przechodzenia do następnej gałęzi ani potrzeby używania `break`.

`case _` pasuje do wszystkiego i zwykle jest ostatnią gałęzią. Jeśli żaden wzorzec nie pasuje, żadna gałąź nie zostanie wykonana.

````python
status = 404
match status:
    case 200:
        message = "OK"
    case 404:
        message = "Nie znaleziono"
    case _:
        message = "Inny status"
print(message)
````

#### Rozszerzenie: kilka wartości w jednej gałęzi

Operator `|` łączy alternatywne wzorce. Tutaj każdy z podanych statusów oznacza tę samą odpowiedź.

````python
def describe_status(status):
    match status:
        case 200 | 201:
            return "Sukces"
        case 401 | 403:
            return "Brak dostępu"
        case 404:
            return "Nie znaleziono"
        case _:
            return "Inny status"

print(describe_status(201))
print(describe_status(403))
assert describe_status(201) == "Sukces"
assert describe_status(403) == "Brak dostępu"
assert describe_status(500) == "Inny status"
````

#### Rozszerzenie: dopasowanie struktury i warunek (guard)

Wzorzec `(0, y)` sprawdza pierwszy element i przypisuje drugi do nazwy `y`. Wzorce sekwencji dopasowują tutaj zarówno krotki, jak i listy.

`case (x, y) if x == y` najpierw dopasowuje strukturę, a potem sprawdza dodatkowy warunek. Bardziej szczegółowe przypadki umieszczaj przed ogólnymi. W tym przykładzie współrzędne są liczbami.

````python
def describe_point(point):
    match point:
        case (0, 0):
            return "Początek układu"
        case (0, y):
            return f"Oś Y: {y}"
        case (x, 0):
            return f"Oś X: {x}"
        case (x, y) if x == y:
            return f"Przekątna: {x}, {y}"
        case (x, y):
            return f"Punkt: {x}, {y}"
        case _:
            return "Oczekiwano pary współrzędnych"

print(describe_point((0, 3)))
print(describe_point([2, 2]))
assert describe_point((0, 0)) == "Początek układu"
assert describe_point([2, 2]) == "Przekątna: 2, 2"
assert describe_point((1, 2, 3)) == "Oczekiwano pary współrzędnych"
````

#### Rozszerzenie — ważna pułapka: nazwa przechwytuje wartość

`case expected:` **nie porównuje** z wcześniej zdefiniowaną zmienną `expected`. Przechwytuje wartość i przypisuje ją do tej nazwy. Do porównania użyj literału, np. `case 404:`, albo warunku:

```python
expected = 404
match status:
    case value if value == expected:
        print("Oczekiwany status")
    case _:
        print("Inny status")
```

Dla pojedynczego warunku zwykłe `if` jest zazwyczaj czytelniejsze. `match` przydaje się szczególnie, gdy rozpoznajemy kilka struktur danych.

Źródło: [oficjalny tutorial — match statements](https://docs.python.org/3/tutorial/controlflow.html#match-statements).

`range(stop)` opisuje liczby od `0` do `stop - 1`. `range(start, stop, step)` pozwala podać początek i krok. Dla dodatniego kroku pusty zakres nie wykona ani jednej iteracji.

`range` nie tworzy listy wszystkich liczb. Poniżej używamy `list(...)` tylko do pokazania zawartości.

**Dwa różne zadania:** `range(n)` to liczby od `0` do `n - 1`, a `range(1, n + 1)` — od `1` do `n`. Tę różnicę wykorzystamy w `opt01`.

````python
range_n = 5
print(list(range(range_n)))            # [0, 1, 2, 3, 4]
print(list(range(1, range_n + 1)))     # [1, 2, 3, 4, 5]
print(list(range(0, range_n, 2)))      # [0, 2, 4]
print(list(range(0)))                 # []
assert sum(range(range_n)) == 10
assert sum(range(1, range_n + 1)) == 15
assert list(range(1, 1)) == []
````

### Porównuj to samo zachowanie

Dopuszczamy nazwę osoby jako napis. Znana nazwa daje przypisany opis; nieznana nazwa, także pusty napis, daje `"Unknown person"`. Obie funkcje zwracają napis, a wypisanie wyniku odbywa się poza nimi.

Dzięki `else` i `dict.get()` oba warianty mają tę samą obsługę brakującej nazwy. Bez niej `if/elif` mógłby nic nie zrobić, a odczyt `dict[name]` zgłosić `KeyError`. Porównujemy rozwiązania tego samego zadania, zanim zaczniemy rozważać ich koszt.

````python
def describe_job_if(name):
    if name == "John":
        return "This is John, he is an artist"
    elif name == "Ted":
        return "This is Ted, he is an engineer"
    elif name == "Kennedy":
        return "This is Kennedy, he is a teacher"
    else:
        return "Unknown person"

print(describe_job_if(name))
````

````python
name_job_dict = {
    "John": "This is John, he is an artist",
    "Ted": "This is Ted, he is an engineer",
    "Kennedy": "This is Kennedy, he is a teacher",
}

def describe_job_dict(name):
    return name_job_dict.get(name, "Unknown person")

print(describe_job_dict(name))
for candidate in ("John", "Ted", "Kennedy", "Alex", ""):
    assert describe_job_if(candidate) == describe_job_dict(candidate)
print(describe_job_dict("Alex"))
````

````python
# Rozszerzenie: grupowanie i setdefault — do powrotu po kolekcjach.
# Lista par zachowuje powtórzenia kluczy; dict nadpisałby pierwsze a.
data = [("a", 1), ("b", 2), ("a", 2)]

grouped = {}
for key, value in data:
    if key in grouped:
        grouped[key].append(value)
    else:
        grouped[key] = [value]
print(grouped)

grouped_with_default = {}
for key, value in data:
    group = grouped_with_default.setdefault(key, [])
    group.append(value)
print(grouped_with_default)
assert grouped == grouped_with_default == {"a": [1, 2], "b": [2]}
````
