### Truthiness a dokładna wartość

- `if data:` sprawdza niepustość zwykłej kolekcji.
- `if not data:` sprawdza jej pustość. `len(data) == 0` też jest poprawne, ale zwykle mniej zwięzłe.
- `if value:` sprawdza wartość logiczną, nie to, czy obiekt jest dokładnie `True`.
- `value is True` ma sens, gdy potrzebujemy odróżnić `True` np. od `1`. Nie stosuj `value == True` jako domyślnego zapisu warunku.

Te reguły dotyczą omawianych kolekcji Pythona; np. tablica NumPy wymaga osobnego określenia, co ma oznaczać warunek.

## Range

```python
for arguments in ((3,), (5, 10), (2, 12, 3), (-7, -30, -5)):
    values = range(*arguments)
    print(f"range{arguments}: {list(values)}")
assert list(range(1, 3)) == [1, 2]
```

```python
v = range(10)
print(type(v))
print(v)
```

### range jest sekwencją, nie listą ani generatorem

Przechowuje początek, koniec i krok, zamiast materializować wszystkie liczby. Można wielokrotnie po nim iterować, indeksować je i sprawdzać długość. `list(range(...))` dopiero tworzy listę.

Końcowa granica jest wyłączona; krok musi być różny od zera.

```python
v = range(10)
print(len(v), v[3], v[-1])
print(list(v[2:8:2]))
assert list(v) == list(v)  # Kolejna iteracja nie zużywa range.
```

### Gdzie szukać szczegółów?

[Dokumentacja range](https://docs.python.org/3.10/library/stdtypes.html#ranges). `help(range)` można uruchomić samodzielnie; pełny raport jest zbyt obszerny na slajd.

### Płytka kopia a zagnieżdżone dane

`list.copy()` tworzy nową listę z tymi samymi referencjami do elementów. Zagnieżdżone listy pozostają współdzielone. To rozwinięcie przykładu `alias = original` z `part01`.

```python
rows = [[1, 2], [3, 4]]
shallow = rows.copy()
shallow[0].append(99)
print(rows, shallow)
assert shallow is not rows
assert shallow[0] is rows[0]
```

### Kopia elementów — jeśli to jest potrzebne

Dla listy list liczb możemy skopiować każdy wiersz. Przy bardziej złożonych strukturach istnieje `copy.deepcopy()` — decyzja zależy od tego, jakie obiekty chcemy współdzielić.

```python
independent = [row.copy() for row in rows]
independent[0].append(100)
assert 100 not in rows[0]
print(rows, independent)
```

### Metody list: działające przykłady

```python
my_list = [1, 2]
my_list.extend([3, 4])
my_list.insert(1, 99)
my_list.remove(99)
print(my_list)
assert my_list == [1, 2, 3, 4]
```

### append() kontra extend()

`append(x)` dodaje **jeden obiekt**. `extend(iterable)` dodaje **kolejne elementy** obiektu iterowalnego. Obie metody zmieniają listę i zwracają `None`.

```python
appended = [1, 2]
extended = [1, 2]
print(appended.append([3, 4]))  # None
extended.extend([3, 4])
print(appended, extended)
assert appended == [1, 2, [3, 4]]
assert extended == [1, 2, 3, 4]
```

```python
my_list = [3, 1, 3, 2]
print(my_list.count(3))
print(my_list.index(3, 1, len(my_list)))  # Zakres [start, stop).
print(my_list.pop())
my_list.reverse()
print(my_list)
my_list.clear()
assert my_list == []
```

### sorted() kontra list.sort()

`sorted(iterable)` zwraca nową listę; `list.sort()` zmienia listę w miejscu i zwraca `None`. Oba sortowania są stabilne: elementy o równych kluczach zachowują wzajemną kolejność.

```python
original = [3, 1, 2]
ordered = sorted(original)
assert original == [3, 1, 2] and ordered == [1, 2, 3]
result = original.sort()
print(original, result)
assert original == ordered and result is None
```

### 2. Dictionary

### Hashowalność klucza

Klucz musi mieć niezmienny hash podczas swojego życia i spójne porównywanie równości. Liczby i napisy nadają się na klucze; listy, słowniki i zbiory `set` — nie.

Krotka jest hashowalna **tylko wtedy, gdy każdy jej element jest hashowalny**. Do tego wrócimy przy krotkach.

```python
empty = {}
print(type(empty), empty == dict())
```

```python
a = dict(one=1, two=2, three=3)
b = {"one": 1, "two": 2, "three": 3}
print(a == b)
assert a == b
```

### 4. Sets

Zbiór przechowuje unikalne, hashowalne elementy. Nie gwarantuje kolejności iteracji i nie obsługuje indeksowania.

Typowe cele: usuwanie duplikatów, sprawdzanie przynależności i operacje zbiorowe. W CPythonie wyszukiwanie ma średnio koszt O(1), a w najgorszym przypadku O(n).

```python
s = {1, 2, 3}
```

```python
empty_set = set()  # {} tworzy pusty słownik.
set_from_list = set([1, 2, 1, 4, 3])
basket = {"apple", "orange", "apple", "pear", "banana"}
print(set_from_list, len(basket))
print("orange" in basket, "crabgrass" in basket)
for fruit in sorted(basket):  # Jawne sortowanie dla przewidywalnego pokazu.
    print(fruit)
assert len(basket) == 4
```

```python
a = set("mississippi")
a.add("r")
a.remove("m")        # KeyError, jeśli elementu nie ma.
a.discard("x")       # Brak błędu dla nieobecnego elementu.
removed = a.pop()     # Dowolny element; nie przewidujemy którego.
print(removed, sorted(a))
a.clear()
assert len(a) == 0
```

```python
a = set("abracadabra")
b = set("alacazam")
print("Różnica:", sorted(a - b))
print("Suma:", sorted(a | b))
print("Część wspólna:", sorted(a & b))
print("Różnica symetryczna:", sorted(a ^ b))
assert a & b == {"a", "c"}
```

### zip() kończy na najkrótszym wejściu

Czwarte pytanie nie zostało użyte. `zip()` zwraca iterator i domyślnie nie zgłasza różnicy długości.

Gdy równa długość jest wymaganiem, użyj `strict=True` (Python 3.10+). Błąd pojawia się **podczas iteracji**, gdy różnica zostanie wykryta.

```python
# Celowy ValueError: pytania i odpowiedzi mają różne długości.
list(zip(questions, answers, strict=True))
```

```python
pairs = list(zip(questions[:3], answers, strict=True))
assert len(pairs) == 3
print(pairs)
```

### Rozpakowanie par

`zip(*pairs)` przekazuje kolejne pary jako osobne argumenty. Wynikiem rozpakowania są krotki. Dla pustego `pairs` nie powstaną dwie kolumny — taki przypadek wymaga osobnej obsługi.

```python
cars = [("Aston", "GPS"), ("Audi", "Car Repair"), ("McLaren", "Sound kit")]
brands, accessories = zip(*cars)
print(brands)
print(accessories)
```

### Comprehensions

**List comprehensions** provide a concise way to create lists. Common applications are to make new lists where each element is the result of some operations applied to each member of another sequence or iterable, or to create a subsequence of those elements that satisfy a certain condition.

```python
squares = []
for x in range(10):
    squares.append(x**2)

squares
```

```python
squares_by_comprehension = [x * x for x in range(10)]
assert squares_by_comprehension == squares
print(squares_by_comprehension)
```

### Przekształcenie każdego elementu

Schemat: `[expression for item in iterable]`. Wyrażenie po lewej określa element wyniku; część `for` określa źródło danych.

```python
words = ["Python", "LIST", "code", "collections"]
lowercase = [word.lower() for word in words]
print(lowercase)
```

### Filtrowanie a wybór wartości

`if` **po części for** filtruje elementy. `if/else` **przed częścią for** wybiera wartość wyniku, zachowując każdy element wejścia.

```python
numbers = [-2, -1, 0, 1, 2]
positive = [x for x in numbers if x > 0]
labels = ["dodatnia" if x > 0 else "pozostała" for x in numbers]
print(positive, labels)
assert len(positive) == 2 and len(labels) == len(numbers)
```

```python
powers = [(x, x ** 2, x ** 3) for x in range(5)]
pairs = [(i, j) for i in range(4) for j in range(i)]
print(powers)
print(pairs)
```

```python
filtered_pairs = [(i, j) for i in range(5) if i > 2 for j in range(i)]
print(filtered_pairs)
```

```python
combinations = [(x, y) for x in [1, 2, 3] for y in [3, 1, 4] if x != y]
print(combinations)
```

it’s equivalent to

```python
combs = []
for x in [1, 2, 3]:
    for y in [3, 1, 4]:
        if x != y:
            combs.append((x, y))
assert combs == combinations
print(combs)
```

### Porównania leksykograficzne

```python
print((1, 2, 3) < (1, 2, 4))
print([1, 2] < [1, 2, -1])  # Wspólny początek, krótsza lista jest mniejsza.
print((1, 2, 3) == (1.0, 2.0, 3.0))
```

Porównanie uporządkowanych sekwencji tego samego typu sprawdza elementy od początku; pierwsza nierówna para rozstrzyga wynik. Elementy muszą obsługiwać wymagane porównanie.

Porządkowanie zbiorów przez `<` oznacza **właściwy podzbiór**, a nie porównanie leksykograficzne. Nie przenoś tej reguły na wszystkie kolekcje.

### Czytelność comprehensions

Używaj ich do budowania kolekcji. Jeśli potrzebujesz efektów ubocznych, `break`, obsługi błędów lub wielu poziomów warunków, zwykła pętla bywa czytelniejsza.

Lista składana buduje wszystkie wyniki od razu. Wyrażenie generatorowe dostarcza je kolejno — porównaliśmy tę różnicę w `opt01`. Zapis `(x * x for x in range(n))` tworzy generator, nie krotkę.

```python
generator = (x * x for x in range(4))
print(list(generator))
print(list(generator))  # Zużyty iterator: [].
assert list(generator) == []
```

## Koszt operacji: dlaczego wybór kolekcji ma znaczenie?

| Operacja w CPythonie | Typowy koszt |
|---|---|
| `len(list)` i indeksowanie listy | O(1) |
| `append()` do listy | O(1) amortyzowane |
| `x in list` | O(n) |
| Wstawianie/usuwanie na początku listy | O(n) |
| Kopia listy | O(n) |
| `key in dict`, `x in set` | O(1) średnio; O(n) w najgorszym przypadku |

`n` to liczba elementów. Zakładamy stały koszt hashowania i porównania; duże liczby czy długie napisy mogą mieć dodatkowy koszt.

### Złożoność a pomiar

O(n) opisuje wzrost kosztu, a nie liczbę sekund. Małe dane, stałe narzuty i rodzaj elementów wpływają na rzeczywisty czas.

Amortyzowane O(1) dla `append()` oznacza niski średni koszt w serii operacji; pojedyncza realokacja może kosztować więcej.

### Ćwiczenie: dobierz kolekcję i sprawdź rozwiązanie

Dane: `visits = ["Ada", "Jan", "Ada", "Ola", "Jan"]`.

1. Zbuduj zbiór unikalnych osób i słownik liczby wizyt.
2. Zapisz listę osób mających co najmniej dwie wizyty; ustal przewidywalną kolejność.
3. Sprawdź rozwiązanie asercjami dla danych powyżej, pustej listy i jednej osoby.
4. Uzasadnij dobór kolekcji i koszt przetwarzania. Który wariant wybrać do wielu zapytań o obecność?

Najpierw napisz rozwiązanie pętlą; comprehensions zastosuj tam, gdzie poprawiają czytelność.

### Przykładowe rozwiązanie — po dyskusji

```python
def summarize_visits(visits):
    counts = {}
    for person in visits:
        counts[person] = counts.get(person, 0) + 1
    frequent = sorted(person for person, count in counts.items() if count >= 2)
    return set(visits), counts, frequent

assert summarize_visits(["Ada", "Jan", "Ada", "Ola", "Jan"]) == (
    {"Ada", "Jan", "Ola"}, {"Ada": 2, "Jan": 2, "Ola": 1}, ["Ada", "Jan"]
)
assert summarize_visits([]) == (set(), {}, [])
assert summarize_visits(["Ada"]) == ({"Ada"}, {"Ada": 1}, [])
```

## Czy krótszy zapis jest szybszy?

Stosujemy procedurę z `opt01`: te same wymagania → zgodne wyniki → powtarzany pomiar → interpretacja.

Wszystkie trzy warianty budują **listę Pythonowych liczb całkowitych** `x² + x³` dla `x` od `0` do `n - 1`, z tym samym zapisem mnożeń. Wariant słownikowy tworzy dodatkowe mapowanie, którego nie wymaga wynik.

Narzędzie pomiarowe jest przygotowaną demonstracją. Interpretujemy porównanie; szczegóły funkcji jako argumentów wrócą na dalszych zajęciach.

```python
import sys
import timeit
from statistics import median

print(sys.version)
print(sys.executable)

def values_loop(n):
    result = []
    for x in range(n):
        result.append(x * x + x * x * x)
    return result
```

```python
def values_list(n):
    return [x * x + x * x * x for x in range(n)]

def values_dict(n):
    mapping = {x: x * x + x * x * x for x in range(n)}
    return list(mapping.values())

methods = {"pętla": values_loop, "lista składana": values_list,
           "słownik i lista": values_dict}
for n in (0, 1, 10, 10_000):
    expected = [x ** 2 + x ** 3 for x in range(n)]
    assert all(method(n) == expected for method in methods.values())
```

### Pomiar: te same dane i cały wynik

`timeit` korzysta z zegara wydajności. Kalibrujemy liczbę wykonań, powtarzamy pomiar i pokazujemy medianę oraz zakres. Każda próba obejmuje utworzenie całej listy.

Pusta pętla nie rozwiązuje zadania, więc nie przedstawiamy jej jako konkurencyjnego wariantu. Pomiar czasu nie jest pomiarem szczytu pamięci.

[Dokumentacja timeit](https://docs.python.org/3.10/library/timeit.html).

```python
def measure_methods(methods, sizes):
    rows = []
    for n in sizes:
        for label, method in methods.items():
            timer = timeit.Timer(lambda method=method, n=n: method(n))
            loops, _ = timer.autorange()
            samples = [elapsed / loops for elapsed in timer.repeat(3, loops)]
            rows.append((n, label, median(samples), min(samples), max(samples)))
    return rows
```

```python
sizes = (10, 1_000, 10_000)
rows = measure_methods(methods, sizes)
for n, label, typical, fastest, slowest in rows:
    print(f"n={n:>5}, {label:<16}: {typical * 1e6:>9.2f} µs "
          f"[{fastest * 1e6:.2f}, {slowest * 1e6:.2f}]")
```

### Wykres: czas a rozmiar danych

Osie logarytmiczne ułatwiają porównanie kilku rzędów wielkości. Linie pokazują mediany z bieżącego uruchomienia; interpretuj je razem z zakresem pomiarów.

Matplotlib jest potrzebny tylko do wizualizacji. W Notebooku 7 i RISE używamy statycznego backendu `inline`.

Zależności tego wykładu (`matplotlib` i NumPy) są zapisane w `pyproject.toml` i `poetry.lock`. Podstawowe przykłady kolekcji korzystają z biblioteki standardowej.

```python
import matplotlib.pyplot as plt
%matplotlib inline

fig, ax = plt.subplots(figsize=(9, 3.5), layout="constrained")
for label in methods:
    selected = [(n, typical) for n, name, typical, _, _ in rows if name == label]
    ax.plot([n for n, _ in selected], [time * 1e6 for _, time in selected],
            marker="o", label=label)
ax.set(xscale="log", yscale="log", xlabel="Liczba elementów n",
       ylabel="Mediana czasu [µs]", title="Ten sam wynik: lista liczb całkowitych")
ax.legend()
ax.grid(True, alpha=0.25)
plt.close(fig)  # Wyświetlimy wykres na osobnym slajdzie.
```

### Rozszerzenie opcjonalne: gdzie pasuje NumPy?

NumPy operuje na jednorodnych tablicach, a nie na listach dowolnych obiektów. Poniżej świadomie wybieramy `float64` i porównujemy **listy liczb zmiennoprzecinkowych**.

Do kosztu NumPy wliczamy utworzenie tablicy, obliczenia i konwersję do listy. W rzeczywistym potoku, który dalej pracuje na tablicach, konwersja może być zbędna.

Dla zakresu poniżej wyniki są dokładnie reprezentowalne; przy większych liczbach trzeba ponownie ustalić wymagania dokładności. NumPy i dtype omówimy szczegółowo później.

```python
import numpy as np

def values_float(n):
    return [float(x * x + x * x * x) for x in range(n)]

def values_numpy_float(n):
    x = np.arange(n, dtype=np.float64)
    return (x * x + x * x * x).tolist()

for n in (0, 1, 10, 10_000):
    assert values_numpy_float(n) == values_float(n)
print("Warianty float64 zgodne w sprawdzonym zakresie")
```

### Opcjonalny pomiar NumPy

Uruchom tę część tylko, gdy NumPy jest dostępny. Porównujemy tę samą reprezentację końcowego wyniku; nie mieszamy tego rankingu z wcześniejszym porównaniem list liczb całkowitych.

```python
float_methods = {"Python float": values_float, "NumPy → lista": values_numpy_float}
float_rows = measure_methods(float_methods, (10, 10_000))
for n, label, typical, fastest, slowest in float_rows:
    print(f"n={n:>5}, {label:<14}: {typical * 1e6:.2f} µs "
          f"[{fastest * 1e6:.2f}, {slowest * 1e6:.2f}]")
```
